```python
from pokerkit import HandHistory
import duckdb
import pandas as pd
```

```python
path = "../../data/raw/ps200_1.phhs"
hands = []

with open(path, "rb") as file:
    hhs = HandHistory.load_all(file)
    for hand in hhs:
        hands.append(hand)

print(f'hand size: {len(hands)}')
```

```python
# extract 
df = pd.DataFrame(hands)

# load into duckDB
results = duckdb.sql("SUMMARIZE df")

results
```
